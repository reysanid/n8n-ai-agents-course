```python
import gradio as gr
import requests
```

```python
def summarize_text(text, api_key):
    if len(text.strip()) < 10:
        return "خطا: متن ورودی خیلی کوتاه است"
    if not api_key.strip():
        return "خطا: اگر کلید خالی بود"
    headers = {"Authorization": f"Bearer {api_key}"}
    data = {"model": "meta-llama/llama-3.2-3b-instruct:free","messages": [{"role": "user", "content": f"متن زیر را به فارسی خلاصه کن:\n{text}"}]}
    response = requests.post("https://openrouter.ai/api/v1/chat/completions", 
    headers=headers, json=data)
    result = response.json()
    summary = result["choices"][0]["message"]["content"]
    return summary
```

```python
app = gr.Interface(
    fn=summarize_text,
    inputs=[gr.Textbox(label="متن ورودی"), gr.Textbox(label="API Key", type="password")],
    outputs=gr.Textbox(label="خلاصه"),
    title="خلاصه ساز متن با هوش مصنوعی")
app.launch()
```
